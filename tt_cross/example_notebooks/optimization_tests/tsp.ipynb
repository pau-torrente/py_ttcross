```python
import numpy as np
import numba as nb

from src.regular_tt_cross.dmrg_cross import greedy_cross, ttrc
```

```python
def tsp_cost_function(x: np.ndarray) -> np.ndarray:
    cost = 0

    for i in range(1, x.shape[0] - 1, 2):
        for j in range(0, x.shape[0], 2):
            cost += x[i] * x[j]

    return cost
```

```python
tsp_cost_function(np.array([1, 1] * 10))
```

```python
ttrc_interpolator = ttrc(
    tsp_cost_function,
    20,
    np.array([[-1, 1] for _ in range(20)]),
    1e-12,
    1e-14,
    20,
    3,
    10,
    pivot_initialization="random",
)
```

```python
ttrc_interpolator.run()
```

```python
ttrc_interpolator.eval(np.array([1, 1] * 10))
```
